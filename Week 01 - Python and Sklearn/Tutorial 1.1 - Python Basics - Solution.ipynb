## Exercise 1: Store reactor conditions

## Exercise 13: Write a temperature conversion function

You need to convert temperatures often. Make a reusable function.

```python
def celsius_to_kelvin(celsius):
    # Convert celsius to kelvin and return the result
    return celsius + 273.15

# Test with the reactor's current temperature
temp_K = celsius_to_kelvin(reactor_data['temperature'])
print(f"Reactor temperature: {temp_K} K")
```

## Exercise 14: Calculate moles using ideal gas law

Calculate how many moles of gas are in the reactor using: n = PV / RT

Where R = 8.314 (gas constant), P is in Pa, T is in K

```python
def calculate_moles(pressure_Pa, volume, temperature_K):
    R = 8.314
    # Calculate moles using: n = PV / RT
    moles = (pressure_Pa * volume) / (R * temperature_K)
    return moles

# Use reactor_data values (remember to convert pressure to Pa and temp to K)
P = reactor_data['pressure'] * 100000  # Convert bar to Pa
V = reactor_data['volume']
T = celsius_to_kelvin(reactor_data['temperature'])

n = calculate_moles(P, V, T)
print(f"Moles in {reactor_name}: {n:.2f} mol")
```

## Exercise 15: Check if reactor is in safe operating range

Write a function to verify the reactor temperature is safe.

```python
def is_safe_temperature(temp, min_temp, max_temp):
    # Return True if temp is between min_temp and max_temp
    return min_temp <= temp <= max_temp

# Check if R-101 is safe (safe range: 80-90°C)
safe = is_safe_temperature(reactor_data['temperature'], 80, 90)
print(f"{reactor_name} temperature safe: {safe}")
```

## Exercise 16: Calculate density from reactor data

The reactor contains 1200 kg of material. Calculate its density.

```python
def calculate_density(mass, volume):
    # Calculate and return density = mass / volume
    return mass / volume

mass = 1200  # kg
density = calculate_density(mass, reactor_data['volume'])
print(f"{reactor_name} density: {density:.2f} kg/m³")
```

## Exercise 17: Calculate conversion for a reaction

Measure inlet and outlet concentrations to find reaction conversion.

```python
def calculate_conversion(inlet, outlet):
    # Calculate conversion = (inlet - outlet) / inlet * 100
    conversion = (inlet - outlet) / inlet * 100
    return conversion

# R-101 has inlet concentration 3.0 mol/L, outlet 0.6 mol/L
conv = calculate_conversion(3.0, 0.6)
print(f"{reactor_name} conversion: {conv}%")
```

## Exercise 18: Monitor multiple reactors

Your plant has three reactors. Store their data and calculate average temperature.

```python
# Create a list with three reactor dictionaries
reactors = [
    {"name": "R-101", "temperature": 85.5, "pressure": 3.5},
    {"name": "R-102", "temperature": 87.2, "pressure": 3.8},
    {"name": "R-103", "temperature": 86.1, "pressure": 3.6}
]

# Calculate average temperature across all reactors
total_temp = 0
for reactor in reactors:
    total_temp += reactor['temperature']  # Add this reactor's temperature to total_temp

avg_temperature = total_temp / len(reactors)  # Divide total by number of reactors

print(f"Average temperature across all reactors: {avg_temperature:.2f} °C")
```

## Exercise 19: Find the hottest reactor

Identify which reactor has the highest temperature.

```python
# Find the maximum temperature
max_temp = 0
hottest_reactor = ""

for reactor in reactors:
    if reactor['temperature'] > max_temp:
        max_temp = reactor['temperature']  # Update max_temp
        hottest_reactor = reactor['name']  # Update hottest_reactor name

print(f"Hottest reactor: {hottest_reactor} at {max_temp} °C")
```
