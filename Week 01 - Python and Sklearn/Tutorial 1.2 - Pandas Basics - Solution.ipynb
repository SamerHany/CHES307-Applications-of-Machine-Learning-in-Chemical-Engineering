## Exercise 7: Filter data for a specific reactor

Get all data for reactor R-101.

```python
# Filter for R-101 only
r101_data = df[df['Reactor_ID'] == 'R-101']

print(f"R-101 has {len(r101_data)} measurements")
print(r101_data.head())
```

## Exercise 8: Find high temperature conditions

Identify when any reactor exceeded 88°C.

```python
# Filter for temperature > 88
high_temp = df[df['Temperature_C'] > 88]

print(f"Found {len(high_temp)} high temperature readings")
print(high_temp[['Timestamp', 'Reactor_ID', 'Temperature_C']].head())
```

## Exercise 9: Find reactors under maintenance

Identify all maintenance periods.

```python
# Filter for Status == 'maintenance'
maintenance = df[df['Status'] == 'maintenance']

print(f"Maintenance records: {len(maintenance)}")
print(maintenance[['Timestamp', 'Reactor_ID', 'Status']])
```

## Exercise 10: Find high temperature in specific reactor

Check if R-101 had any readings above 87°C.

```python
# Filter for R-101 AND temperature > 87
r101_high_temp = df[(df['Reactor_ID'] == 'R-101') & (df['Temperature_C'] > 87)]

print(f"R-101 high temperature events: {len(r101_high_temp)}")
print(r101_high_temp[['Timestamp', 'Temperature_C', 'Pressure_bar']].head())
```

## Exercise 11: Find optimal operating conditions

Find records where conversion is above 80% and status is operational.

```python
# Filter for Conversion_pct > 80 AND Status == 'operational'
optimal = df[(df['Conversion_pct'] > 80) & (df['Status'] == 'operational')]

print(f"Optimal operating records: {len(optimal)}")
print(f"That's {len(optimal)/len(df)*100:.1f}% of all data")
```

## Exercise 12: Calculate average conditions for R-101

Find the average temperature, pressure, and conversion for reactor R-101.

```python
# Calculate average temperature for R-101
avg_temp_r101 = r101_data['Temperature_C'].mean()

# Calculate average pressure for R-101
avg_pressure_r101 = r101_data['Pressure_bar'].mean()

# Calculate average conversion for R-101
avg_conversion_r101 = r101_data['Conversion_pct'].mean()

print(f"R-101 Average Conditions:")
print(f"  Temperature: {avg_temp_r101:.2f} °C")
print(f"  Pressure: {avg_pressure_r101:.2f} bar")
print(f"  Conversion: {avg_conversion_r101:.2f} %")
```

## Exercise 13: Add temperature in Kelvin

Create a new column with temperature in Kelvin.

```python
# Add new column: Temperature_K = Temperature_C + 273.15
df['Temperature_K'] = df['Temperature_C'] + 273.15

print(df[['Reactor_ID', 'Temperature_C', 'Temperature_K']].head())
```

## Exercise 14: Categorize conversion efficiency

Create a column that labels conversion as 'High' (>82%), 'Medium' (75-82%), or 'Low' (<75%).

```python
def categorize_conversion(conversion):
    if conversion > 82:
        return 'High'
    elif conversion >= 75:
        return 'Medium'
    else:
        return 'Low'

# Apply the function to create new column
df['Efficiency'] = df['Conversion_pct'].apply(categorize_conversion)

print(df[['Reactor_ID', 'Conversion_pct', 'Efficiency']].head(10))
```

## Exercise 15: Count efficiency categories

How many readings fall into each efficiency category?

```python
# Count values in Efficiency column
efficiency_counts = df['Efficiency'].value_counts()

print("Efficiency distribution:")
print(efficiency_counts)
```

## Exercise 16: Check for missing data

Identify how many missing values are in each column.

```python
# Count missing values per column
missing = df.isnull().sum()

print("Missing values:")
print(missing)
```

## Exercise 17: Fill missing temperature values

Replace missing temperature values with the average temperature.

```python
# Calculate mean temperature
mean_temp = df['Temperature_C'].mean()

# Fill missing values in Temperature_C
df['Temperature_C'].fillna(mean_temp, inplace=True)

print(f"Filled missing temperatures with {mean_temp:.2f}°C")
print(f"Missing temperatures now: {df['Temperature_C'].isnull().sum()}")
```

## Exercise 18: Sort by conversion

Find the top 10 best performing readings.

```python
# Sort by Conversion_pct in descending order and get top 10
top_10 = df.sort_values('Conversion_pct', ascending=False).head(10)

print("Top 10 conversion readings:")
print(top_10[['Timestamp', 'Reactor_ID', 'Conversion_pct', 'Temperature_C', 'Pressure_bar']])
```

## Exercise 19: Group by reactor and calculate averages

Calculate average temperature, pressure, and conversion for each reactor.

```python
# Group by Reactor_ID and calculate mean for only numeric columns of interest
reactor_averages = df.groupby('Reactor_ID')[['Temperature_C', 'Pressure_bar', 'Conversion_pct']].mean()

print("Average conditions by reactor:")
print(reactor_averages.round(2))
```

## Exercise 20: Find the best reactor

Which reactor has the highest average conversion?

```python
# Get average conversion by reactor
avg_conversion_by_reactor = df.groupby('Reactor_ID')['Conversion_pct'].mean()

# Find the reactor with maximum average conversion
best_reactor = avg_conversion_by_reactor.idxmax()
best_conversion = avg_conversion_by_reactor.max()

print(f"Best performing reactor: {best_reactor}")
print(f"Average conversion: {best_conversion:.2f}%")
```

## Exercise 21: Filter operational data only

Create a clean dataset with only operational readings for further analysis.

```python
# Filter for operational status only
operational_data = df[df['Status'] == 'operational']

print(f"Total records: {len(df)}")
print(f"Operational records: {len(operational_data)}")
print(f"Data availability: {len(operational_data)/len(df)*100:.1f}%")
```

## Exercise 22: Calculate pressure range by reactor

Find the minimum and maximum pressure for each reactor.

```python
# Group by reactor and calculate min and max pressure
pressure_range = df.groupby('Reactor_ID')['Pressure_bar'].agg(['min', 'max'])

# Calculate range
pressure_range['range'] = pressure_range['max'] - pressure_range['min']

print("Pressure ranges by reactor:")
print(pressure_range.round(2))
```
