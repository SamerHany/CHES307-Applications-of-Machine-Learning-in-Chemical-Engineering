<h3> 1. What is a DataFrame? </h3>

A DataFrame is like a table or spreadsheet in Python. It has:
- **Rows**: Each row represents one record (like one student, one product, etc.)
- **Columns**: Each column represents one feature or property

Think of it as an Excel spreadsheet you can work with in Python!

```python
# Import pandas
import pandas as pd
import numpy as np

print("Pandas version:", pd.__version__)
```

<hr id="2">
<h3> 2. Creating DataFrames </h3>

Let's create DataFrames using simple examples.

```python
# Example 1: Student grades
students = {
    'Name': ['Alice', 'Bob', 'Charlie', 'Diana', 'Eve'],
    'Age': [20, 21, 20, 22, 21],
    'Math': [85, 78, 92, 88, 76],
    'Science': [90, 82, 88, 95, 80]
}

df = pd.DataFrame(students)
print("Student Grades:")
print(df)
```

```python
# Example 2: Products in a store
products = {
    'Product': ['Apple', 'Banana', 'Orange', 'Grapes'],
    'Price': [1.2, 0.5, 0.8, 2.5],
    'Stock': [100, 150, 80, 60]
}

df_products = pd.DataFrame(products)
print("\nStore Products:")
print(df_products)
```

<h3> 3. Viewing Data </h3>

Let's learn how to inspect our data.

<hr id="4">
<h3> 4. Selecting Data </h3>

How to get specific parts of your data.

```python
# Select one column
print("Just the Math grades:")
print(df['Math'])
```

```python
# Select multiple columns
print("\nName and Math grades:")
print(df[['Name', 'Math']])
```

```python
# Select rows by position using iloc
print("\nFirst row:")
print(df.iloc[0])

print("\nFirst 3 rows:")
print(df.iloc[0:3])
```

```python
# Select specific rows and columns
print("\nFirst 3 students, only Name and Math:")
print(df.loc[0:2, ['Name', 'Math']])
```

<hr id="5">
<h3> 5. Filtering Data </h3>

How to find rows that match certain conditions.

```python
# Students who scored above 85 in Math
print("Students with Math > 85:")
high_math = df[df['Math'] > 85]
print(high_math)
```

```python
# Students who are 21 years old
print("\nStudents who are 21:")
age_21 = df[df['Age'] == 21]
print(age_21)
```

```python
# Multiple conditions: Math > 80 AND Science > 85
print("\nStudents with Math > 80 AND Science > 85:")
good_students = df[(df['Math'] > 80) & (df['Science'] > 85)]
print(good_students)
```

```python
# OR condition: Math > 90 OR Science > 90
print("\nStudents with Math > 90 OR Science > 90:")
excellent = df[(df['Math'] > 90) | (df['Science'] > 90)]
print(excellent)
```

<hr id="6">
<h3> 6. Adding and Modifying Columns </h3>

Creating new columns based on existing data.

```python
# Calculate average of Math and Science
df['Average'] = (df['Math'] + df['Science']) / 2
print("Added Average column:")
print(df)
```

```python
# Add a grade based on average
def get_grade(average):
    if average >= 90:
        return 'A'
    elif average >= 80:
        return 'B'
    elif average >= 70:
        return 'C'
    else:
        return 'D'

df['Grade'] = df['Average'].apply(get_grade)
print("\nAdded Grade column:")
print(df)
```

```python
# Add a constant value
df['Year'] = 2024
print("\nAdded Year column:")
print(df)
```

<hr id="7">
<h3> 7. Basic Statistics </h3>

Calculate common statistics for your data.

```python
# Mean (average)
print("Average Math score:", df['Math'].mean())
print("Average Science score:", df['Science'].mean())
```

```python
# Maximum and minimum
print("\nHighest Math score:", df['Math'].max())
print("Lowest Math score:", df['Math'].min())
```

```python
# Sum and count
print("\nTotal of all Math scores:", df['Math'].sum())
print("Number of students:", df['Name'].count())
```

```python
# Standard deviation (spread of data)
print("\nStandard deviation of Math scores:", df['Math'].std().round(2))
```

<hr id="8">
<h3> 8. Grouping and Sorting Data </h3>

First, summarize data by groups. Then sort those summaries or the original table.

```python
# Step 1: Group by Age and calculate average scores.
age_summary = df.groupby('Age')[['Math', 'Science']].mean()
print('Average scores by Age:')
print(age_summary.round(2))
```

```python
# Step 2: Sort the grouped summary by average Math score.
age_summary_sorted = age_summary.sort_values('Math', ascending=False)
print('\nAge groups sorted by average Math:')
print(age_summary_sorted.round(2))
```

```python
# Step 3: Sort the original DataFrame by two columns.
print('\nOriginal data sorted by Age, then Math:')
df_sorted = df.sort_values(['Age', 'Math'], ascending=[True, False])
print(df_sorted[['Name', 'Age', 'Math']])
```

<h3> 9. Handling Missing Data </h3>

Sometimes data has missing values. Let's see how to handle them.

```python
# Create data with missing values
data_with_missing = {
    'Name': ['Frank', 'Grace', 'Henry', 'Iris', 'Jack'],
    'Age': [20, None, 22, 21, 20],
    'Math': [75, 88, None, 92, 85],
    'Science': [80, 85, 90, None, 88]
}

df_missing = pd.DataFrame(data_with_missing)
print("Data with missing values:")
print(df_missing)
```

```python
# Check for missing values
print("\nMissing values per column:")
print(df_missing.isnull().sum())
```

```python
# Drop rows with any missing values
df_dropped = df_missing.dropna()
print("\nAfter dropping rows with missing values:")
print(df_dropped)
```

```python
# Fill missing values with a number
df_filled = df_missing.fillna(0)
print("\nFilled missing values with 0:")
print(df_filled)
```

```python
# Fill missing values with column average
df_filled_mean = df_missing.fillna(df_missing.mean())
print("\nFilled with column averages:")
print(df_filled_mean.round(1))
```
