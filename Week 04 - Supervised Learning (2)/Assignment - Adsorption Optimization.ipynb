# Assignment: Adsorption Process Optimization


## Problem Description

A pilot adsorption unit has been run over several months. Data were collected from operating logs (concentration, temperature, pressure, run order, reactor ID, shift, etc.) and from periodic lab analyses of adsorbent samples that report equilibrium adsorption capacity (mmol/g). The dataset you receive is a consolidation of these logs and lab results.

**Objective**

Find the combination of concentration and temperature within the allowed operating bounds (concentration in [0.2, 2.8] mol/L, temperature in [295, 345] K) that **maximizes** equilibrium adsorption capacity.

## Tasks

1. Load the assignment data and explore it (inspect, summarize, visualize relationships with the target).
2. Select features and target; use only variables that physically affect adsorption.
3. Split data, scale features, and build a regression model suitable for this data. Evaluate it (e.g. RMSE, R² on a test set).
4. Using the trained model, find the operating conditions (concentration and temperature) within the given bounds that maximize predicted adsorption capacity.
5. Report the optimal concentration, temperature, and predicted capacity; add a short comment on limitations or assumptions.
