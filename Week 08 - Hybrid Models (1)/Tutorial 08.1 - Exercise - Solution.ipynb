## 1. Define the system parameters and exact solution (for model evaluation and visualization)

#### Import necessary Libraries (*this is complete*)

```python
import torch
import torch.nn as nn
import numpy as np
import matplotlib.pyplot as plt
```

#### Define system parameters (*this is complete*)

```python
# System model
def exact_solution(d, w0, t):
    w = np.sqrt(w0**2 - d**2)
    phi = np.arctan(-d/w)
    A = 1/(2*np.cos(phi))
    cos = torch.cos(phi+w*t)
    exp = torch.exp(-d*t)
    u = exp*2*A*cos
    return u


# Create noisy observations
torch.manual_seed(123)
d = 2
w0 = 20
k = w0**2
t_obs = torch.rand(40).view(-1, 1) * 0.4
u_obs = exact_solution(d, w0, t_obs) + 0.04 *torch.randn_like(t_obs)

t_test = torch.linspace(0, 1, 300).view(-1, 1)
u_exact = exact_solution(d, w0, t_test)

# plot the data
def plot_observed_data():
    plt.scatter(t_obs, u_obs, label='Observed data', color='red')
    plt.plot(t_test, u_exact, label='Exact solution', color='grey', alpha=0.6)
    plt.legend()
    plt.show()

plot_observed_data()
```

## 2. Neural Network model

```python
# create neural network class
class NN(nn.Module):
    def __init__(self, n_input, n_output, n_hidden, n_layers, act_fn=nn.Tanh()):
        super().__init__()
        self.act_fn = act_fn

        # create the first hidden layer connected to the inputs
        self.nn_start = nn.Sequential(nn.Linear(n_input, n_hidden), act_fn)

        # create the rest of the hidden layers
        self.nn_hidden = nn.Sequential(*[nn.Sequential(nn.Linear(n_hidden, n_hidden), act_fn) for _ in range(n_layers-1)])

        # create the output layer
        self.nn_end = nn.Linear(n_hidden, n_output)


    # define the forward pass of the network
    def forward(self, x):
        x = self.nn_start(x)
        x = self.nn_hidden(x)
        x = self.nn_end(x)
        return x

    # define the fit method
    def fit(self, X, y, n_epochs, lr=0.01):

        # define the optimizer and the loss function
        optimizer = torch.optim.Adam(self.parameters(), lr=lr)
        loss_fn = nn.MSELoss()

        # train the model
        for epoch in range(n_epochs+1):
            # 1. reset the gradients back to zero
            optimizer.zero_grad()

            # 2. compute the output prediction
            y_pred = self(X)

            # 3. compute the loss
            loss = loss_fn(y_pred, y)

            # 4. backpropagate the loss
            loss.backward()

            # 5. update the weights
            optimizer.step()

            # plot the result as training progresses
            if epoch % 2000 == 0:
                y_pred = self.forward(t_test).detach()
                plt.figure(figsize=(12, 4))
                plt.title(f'Train step {epoch}')
                plt.plot(t_test[:,0], y_pred[:,0], label='NN solution', color='blue', alpha=0.6)
                plot_observed_data()


# TODO: create an instance of the NN class and train it
# create a neural network instance
model_nn = NN(1, 1, 10, 2)

# train the NN
model_nn.fit(t_obs, u_obs, 10000, lr=0.001)

```

## 3. PINN model

```python
# create neural network class
class PINN(nn.Module):
    # define the class constructor
    def __init__(self, n_input, n_output, n_hidden, n_layers, act_fn=nn.Tanh()):
        super().__init__()
        # TODO: define the network architecture
        self.act_fn = act_fn

        # create the first hidden layer connected to the inputs
        self.nn_start = nn.Sequential(nn.Linear(n_input, n_hidden), act_fn)

        # create the rest of the hidden layers
        self.nn_hidden = nn.Sequential(
            *[nn.Sequential(nn.Linear(n_hidden, n_hidden), act_fn) for _ in range(n_layers-1)]
            )

        # create the output layer
        self.nn_end = nn.Linear(n_hidden, n_output)

    # define the forward pass of the network
    def forward(self, x):
        # TODO: define the forward pass
        x = self.nn_start(x)
        x = self.nn_hidden(x)
        x = self.nn_end(x)
        return x


    # define the fit method
    def fit(self, X, y, n_epochs, lr=0.01):

        # TODO: define the optimizer and loss function
        optimizer = torch.optim.Adam(self.parameters(), lr=lr)
        loss_fn = nn.MSELoss()

        # TODO: define the physical parameters
        x_phys = torch.linspace(0,1,30).view(-1,1).requires_grad_(True)
        mu, k = 2*d, w0**2
        lambda_data = 1
        lambda_phys = 0.0001

        # train the model
        for epoch in range(n_epochs + 1):
            # TODO: implement steps for training the model including the physics-informed loss
            # 1. reset the gradients back to zero
            optimizer.zero_grad()

            # 2. compute the output using the trained model
            y_pred = self.forward(X)

            # 3.1. compute data loss
            loss_data = loss_fn(y_pred, y)

            # CHANGE #3: compute the gradients of the physical loss
            # 3.2. compute physics loss
            y_phys = self.forward(x_phys)

            # compute the derivatives of y with respect to x
            # this is done by using the autograd.grad function
            # We use torch.autograd.grad to compute derivatives with respect to the input because 
            # our PINN loss includes terms based on the physical law (an ODE with derivatives of y). 
            # Unlike standard neural nets where we only calculate parameter gradients during backpropagation, 
            # here we need explicit gradients with respect to the input variable x for the physics loss.
            # 'create_graph=True' ensures that the computation graph is kept so we can compute higher order derivatives.
            
            # Compute dy/dx using autograd.grad: this gives us the first derivative of the network output 
            # with respect to input x_phys, required for the ODE residual.
            dx  = torch.autograd.grad(
                y_phys, x_phys, torch.ones_like(y_phys), create_graph=True
            )[0]  # computes dy/dx
            
            # Compute d^2y/dx^2 by differentiating dy/dx again with respect to x_phys.
            # This is needed because our ODE includes a second derivative term. 
            dx2 = torch.autograd.grad(
                dx, x_phys, torch.ones_like(dx), create_graph=True
            )[0]  # computes d^2y/dx^2

            # Put it all together
            loss_phys = loss_fn(dx2 + mu*dx, -1*k*y_phys)

            # 3.3. compute the total loss
            loss = lambda_data*loss_data + lambda_phys*loss_phys

            # 4. backpropagate the loss
            loss.backward()

            # 5. update the weights
            optimizer.step()


            # plot the result as training progresses
            if epoch % 5000 == 0:
                y_pred = self.forward(t_test).detach()
                plt.figure(figsize=(12, 4))
                plt.title(f'Train step {epoch}')
                plt.plot(t_test[:,0], y_pred[:,0], label='PINN solution', color='blue', alpha=0.6)
                plot_observed_data()
                print(f"Physics loss: {loss_phys:.6f}, Data loss: {loss_data:.6f}")

# TODO: create an instance of the PINN class and train it
# create a neural network instance
model_pinn = PINN(1, 1, 10, 2)

# train the NN
model_pinn.fit(t_obs, u_obs, 30000, lr=0.01)
```
